```python
import csv

def parse_geonames(filepath):
    cleaned = []

    with open(filepath, encoding="utf-8") as f:
        reader = csv.reader(f, delimiter="\t")

        for row in reader:
            try:
                name = row[1].strip().lower()
                lat = float(row[4])
                lon = float(row[5])
                feature_code = row[7]
                population = int(row[14]) if row[14] else 0

                # filter only useful place types
                if not feature_code.startswith("PPL") and not feature_code.startswith("ADM"):
                    continue

                cleaned.append({
                    "name": name,
                    "lat": lat,
                    "lon": lon,
                    "feature_code": feature_code,
                    "population": population
                })

            except Exception:
                continue

    return cleaned

cleaned_data = parse_geonames(r"C:\Users\kuchbhe\Desktop\Travelara-ai\backend\tests\pipeline_test\data\cities15000\cities15000.txt")

cleaned_df = pd.DataFrame(cleaned_data)
cleaned_df.head()
```

```python
import pandas as py
from dotenv import load_dotenv
import os

load_dotenv()



CATEGORIES = [
    "restaurants",
    "cafes",
    "tourist attractions",
    "museums",
    "parks"
]

CITIES = [
    ("Delhi", 28.6139, 77.2090),
    ("Mumbai", 19.0760, 72.8777),
    ("Bangalore", 12.9716, 77.5946),
    ("Jaipur", 26.9124, 75.7873),
    ("Goa", 15.2993, 74.1240),
    ("Chandigarh", 30.7333, 76.7794),
    ("Hyderabad", 17.3850, 78.4867),
    ("Kolkata", 22.5726, 88.3639),
    ("Pune", 18.5204, 73.8567),
    ("Ahmedabad", 23.0225, 72.5714),

    # Add global cities for generalization
    ("New York", 40.7128, -74.0060),
    ("London", 51.5074, -0.1278),
    ("Paris", 48.8566, 2.3522),
    ("Tokyo", 35.6762, 139.6503),
    ("Dubai", 25.2048, 55.2708),
    ("Singapore", 1.3521, 103.8198),
    ("Sydney", -33.8688, 151.2093),
    ("Toronto", 43.6532, -79.3832),
    ("Berlin", 52.5200, 13.4050),
    ("Bangkok", 13.7563, 100.5018),
]

import requests
import time
import random

API_KEY = os.getenv("SERP_API_KEY")
BASE_URL = "https://serpapi.com/search"

CALL_DELAY = 2.0  # seconds (safe buffer)
MAX_CALLS = 200

# i/p params:



def build_ll(lat, lon, zoom=14):
    return f"@{lat},{lon},{zoom}z"

def fetch_places(query, lat, lon, start=0):

    params = {
        "engine": "google_maps",
        "q": query,
        "ll": build_ll(lat, lon),
        "start": start,
        "api_key": API_KEY
    }

    response = requests.get(BASE_URL, params=params)

    if response.status_code != 200:
        print("Error:", response.status_code, response.text)
        return None

    return response.json()

seen_place_ids = set()

def filter_new_places(results):
    new_places = []

    for place in results.get("local_results", []):
        pid = place.get("place_id")
        if pid and pid not in seen_place_ids:
            seen_place_ids.add(pid)
            new_places.append(place)

    return new_places

def collect_data():

    all_results = []
    call_count = 0

    for city, lat, lon in CITIES:
        for category in CATEGORIES:

            for start in [0, 20]:  # pagination

                if call_count >= MAX_CALLS:
                    return all_results

                print(f"Fetching {category} in {city}, page {start}")

                results = fetch_places(category, lat, lon, start)

                if not results:
                    continue

                new_places = filter_new_places(results)

                all_results.extend([
                    {
                        "city": city,
                        "category_query": category,
                        "data": place
                    }
                    for place in new_places
                ])

                call_count += 1

                # rate limiting
                time.sleep(CALL_DELAY + random.uniform(0.2, 0.8))

                if len(new_places) < 5:
                    break  # low yield → stop pagination

    return all_results

import json

data = collect_data()

with open("places_raw.json", "w") as f:
    json.dump(data, f, indent=2)

    
```
